```python
results = []     # every evaluated model adds one row here


class KerasWrapper:
    """Makes a Keras model behave like a scikit-learn classifier."""
    def __init__(self, model):
        self.model = model
    def predict_proba(self, X):
        p = self.model.predict(X, verbose=0).ravel()
        return np.column_stack([1 - p, p])
    def predict(self, X):
        return (self.model.predict(X, verbose=0).ravel() > 0.5).astype(int)


def evaluate_model(name, model, X_train, y_train, X_test, y_test):


    if hasattr(model, "layers"):               # a Keras model: wrap it
        model = KerasWrapper(model)

    pred_train = model.predict(X_train)
    pred_test  = model.predict(X_test)

    tn, fp, fn, tp = confusion_matrix(y_test, pred_test, labels=[0, 1]).ravel()

    row = {
        'Model':          name,
        'Train Acc':      round(accuracy_score(y_train, pred_train) * 100, 2),
        'Test Acc':       round(accuracy_score(y_test, pred_test) * 100, 2),
        'Precision':      round(precision_score(y_test, pred_test, zero_division=0) * 100, 2),
        'Recall':         round(recall_score(y_test, pred_test) * 100, 2),
        'Specificity':    round(tn / (tn + fp) * 100, 2),
        'F1':             round(f1_score(y_test, pred_test) * 100, 2),
        'TP': tp, 'FN': fn, 'FP': fp, 'TN': tn,

    }

    results[:] = [r for r in results if r['Model'] != name]   # rerunning replaces the old row
    results.append(row)
    return pd.DataFrame(results)
```

### Evaluating Classification Error

```python
def eval_cat_err(y,yhat):
  m=len(y)
  incorrect = 0
  for i in range (m):
    if yhat[i] != y.iloc[i]:
      incorrect +=1
  cerr = incorrect /m
  return cerr
```

```python
import os
import random

# 1. Set seeds for all libraries
os.environ['PYTHONHASHSEED'] = '11'
random.seed(11)
np.random.seed(11)
tf.random.set_seed(11)

# 2. Force deterministic operations (optional but powerful)
tf.config.experimental.enable_op_determinism()

simple_model = Sequential([
    Dense(6, activation='relu', name='layer1'),
    Dense(4, activation='relu', name='layer2'),
    Dense(1, activation='sigmoid', name='layer3'),
],name = "simple_model")

simple_model.compile(
    loss=tf.keras.losses.BinaryCrossentropy(),
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01)
)
```

```python
start = time.time()
simple_model.fit(x_fit_scaled,y_fit,epochs=150,verbose=0,validation_data=(x_cv_scaled,y_cv),callbacks=[tf.keras.callbacks.EarlyStopping(monitor='val_loss',restore_best_weights=True)])
end = time.time()
print(f"Training time: {end - start:.2f} seconds")
```

```python
os.environ['PYTHONHASHSEED'] = '12'
random.seed(12)
np.random.seed(12)
tf.random.set_seed(12)

# 2. Force deterministic operations (optional but powerful)
tf.config.experimental.enable_op_determinism()
lambdas = [0.0,0.001,0.01, 0.05, 0.1]
models=[None] * len(lambdas)
for i in range(len(lambdas)):
  lambda_ = lambdas[i]
  models[i] = Sequential([
    Dense(6, activation='relu', name='layer1',kernel_regularizer=tf.keras.regularizers.l2(lambda_)),
    Dense(4, activation='relu', name='layer2',kernel_regularizer=tf.keras.regularizers.l2(lambda_)),
    Dense(1, activation='sigmoid', name='layer3'),
])
  models[i].compile(
    loss=tf.keras.losses.BinaryCrossentropy(),
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01)
)

  models[i].fit(
       x_fit_scaled,y_fit,
       epochs=150,verbose=0,validation_data=(x_cv_scaled,y_cv),callbacks=[tf.keras.callbacks.EarlyStopping(monitor='val_loss',restore_best_weights=True)]
   )
  print(f"Finished lambda = {lambda_}")
```

```python
def plot_iterate(lambdas, models, x_train, y_train, x_cv, y_cv):
    train_losses = []
    cv_losses = []
    train_accuracies = []
    cv_accuracies = []

    bce = tf.keras.losses.BinaryCrossentropy()

    for model in models:
        # Predict probabilities
        y_train_pred = model.predict(x_train, verbose=0).flatten()
        y_cv_pred = model.predict(x_cv, verbose=0).flatten()

        # Compute loss
        train_loss = bce(y_train, y_train_pred).numpy()
        cv_loss = bce(y_cv, y_cv_pred).numpy()

        # Compute accuracy
        y_train_class = (y_train_pred >= 0.5).astype(int)
        y_cv_class = (y_cv_pred >= 0.5).astype(int)
        train_acc = accuracy_score(y_train, y_train_class)
        cv_acc = accuracy_score(y_cv, y_cv_class)

        train_losses.append(train_loss)
        cv_losses.append(cv_loss)
        train_accuracies.append(train_acc)
        cv_accuracies.append(cv_acc)

    # Plotting
    plt.figure(figsize=(12, 5))

    # Loss plot
    plt.subplot(1, 2, 1)
    plt.plot(lambdas, train_losses, marker='o', label='Train Loss')
    plt.plot(lambdas, cv_losses, marker='s', label='CV Loss')
    plt.xlabel('Lambda')
    plt.ylabel('Binary Crossentropy Loss')
    plt.title('Loss vs Lambda')
    plt.legend()
    plt.grid(True)

    # Accuracy plot
    plt.subplot(1, 2, 2)
    plt.plot(lambdas, train_accuracies, marker='o', label='Train Accuracy')
    plt.plot(lambdas, cv_accuracies, marker='s', label='CV Accuracy')
    plt.xlabel('Lambda')
    plt.ylabel('Accuracy')
    plt.title('Accuracy vs Lambda')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()
```

```python
plot_iterate(lambdas, models, x_fit_scaled, y_fit, x_cv_scaled, y_cv)
```

```python
from sklearn.base import clone
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import RepeatedStratifiedKFold, cross_validate
from sklearn.metrics import make_scorer, recall_score

cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=10, random_state=RANDOM_STATE)
scoring = {'accuracy': 'accuracy', 'precision': 'precision', 'recall': 'recall',
           'f1': 'f1', 'specificity': make_scorer(recall_score, pos_label=0),
           'roc_auc': 'roc_auc'}

models_cv = {
    'KNN':                 make_pipeline(StandardScaler(), clone(knn_best)),
    'Logistic Regression': make_pipeline(StandardScaler(), clone(lr_best)),
    'Random Forest':       make_pipeline(StandardScaler(), clone(Tuned_rf)),
    'XGBoost':             clone(Tuned_xgb_model),
}

rows = []
for name, m in models_cv.items():
    s = cross_validate(m, x, y, cv=cv, scoring=scoring, n_jobs=-1)
    rows.append({'Model': name,
                 'Accuracy':    round(s['test_accuracy'].mean() * 100, 1),
                 'Precision':   round(s['test_precision'].mean() * 100, 1),
                 'Recall':      round(s['test_recall'].mean() * 100, 1),
                 'Specificity': round(s['test_specificity'].mean() * 100, 1),
                 'F1':          round(s['test_f1'].mean() * 100, 1),
                 'F1 spread ±': round(s['test_f1'].std() * 100, 1),
                 'ROC-AUC':     round(s['test_roc_auc'].mean(), 3)})
print(pd.DataFrame(rows).to_string(index=False))
```
